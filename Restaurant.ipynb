```python
df['Date'] = pd.to_datetime(df['Date'])

df['Hour'] = df['Hour'].astype(int)

numeric_cols = ['Order_Volume', 'Average_Preparation_Time', 'Queue_Length', 'Staff_Count']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Data types converted successfully for 'Date', 'Hour', and other numerical columns.")
print('\nUpdated Data Types:')
display(df.info())
```

```python
missing_values_after_conversion = df.isnull().sum()
print('Missing values after data type conversion:')
display(missing_values_after_conversion[missing_values_after_conversion > 0])


initial_rows = df.shape[0]
df.dropna(subset=['Order_Volume', 'Average_Preparation_Time', 'Queue_Length', 'Staff_Count'], inplace=True)
rows_after_na = df.shape[0]

print(f"\nDropped {initial_rows - rows_after_na} rows with missing values in key operational metrics.")
print(f"Remaining rows: {df.shape[0]}")

print('\nMissing values check after dropping rows:')
display(df.isnull().sum())
```

```python
min_date = df['Date'].min()
max_date = df['Date'].max()
print(f"Date range: {min_date.strftime('%Y-%m-%d')} to {max_date.strftime('%Y-%m-%d')}")

invalid_hours = df[(df['Hour'] < 0) | (df['Hour'] > 23)]
print(f'Number of invalid hours (outside 0-23 range): {invalid_hours.shape[0]}')

if not invalid_hours.empty:
    print("Invalid hours detected. Reviewing these entries...")
    display(invalid_hours)
    df = df[(df['Hour'] >= 0) & (df['Hour'] <= 23)]
    print(f"Dropped {invalid_hours.shape[0]} rows with invalid hours. Remaining rows: {df.shape[0]}")
else:
    print('No invalid hours found.')
```

```python
df['Year'] = df['Date'].dt.year

df['Month'] = df['Date'].dt.month

df['Month_Name'] = df['Date'].dt.month_name()

df['Day_of_Week'] = df['Date'].dt.dayofweek

df['Day_of_Week_Name'] = df['Date'].dt.day_name()

df['Is_Weekend'] = df['Day_of_Week'].apply(lambda x: 1 if x >= 5 else 0)

def get_time_period(hour):
    if 6 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Period'] = df['Hour'].apply(get_time_period)

print("Time features created successfully.")
print("Displaying new time features for verification:")
display(df[['Date', 'Hour', 'Year', 'Month', 'Month_Name', 'Day_of_Week', 'Day_of_Week_Name', 'Is_Weekend', 'Time_Period']].head())
```

```python
epsilon = 1e-6
df['Orders_Per_Employee'] = df['Order_Volume'] / (df['Staff_Count'] + epsilon)

df['Orders_Per_Staff_Hour'] = df['Orders_Per_Employee']

print("Staffing metrics 'Orders_Per_Employee' and 'Orders_Per_Staff_Hour' created successfully.")
print("Displaying new staffing metrics for verification:")
display(df[['Order_Volume', 'Staff_Count', 'Orders_Per_Employee', 'Orders_Per_Staff_Hour']].head())
```

```python
low_demand_threshold = df['Order_Volume'].quantile(0.25)
normal_demand_threshold = df['Order_Volume'].quantile(0.50)
high_demand_threshold = df['Order_Volume'].quantile(0.75)

def classify_demand(order_volume):
    if order_volume <= low_demand_threshold:
        return 'Low Demand'
    elif low_demand_threshold < order_volume <= normal_demand_threshold:
        return 'Normal Demand'
    elif normal_demand_threshold < order_volume <= high_demand_threshold:
        return 'High Demand'
    else:
        return 'Peak Demand'

df['Demand_Category'] = df['Order_Volume'].apply(classify_demand)

print("Demand categories created based on Order_Volume quantiles:")
print(f"  - Low Demand: Order_Volume <= {low_demand_threshold:.2f}")
print(f"  - Normal Demand: {low_demand_threshold:.2f} < Order_Volume <= {normal_demand_threshold:.2f}")
print(f"  - High Demand: {normal_demand_threshold:.2f} < Order_Volume <= {high_demand_threshold:.2f}")
print(f"  - Peak Demand: Order_Volume > {high_demand_threshold:.2f}")
print("Displaying demand categories for verification:")
display(df[['Order_Volume', 'Demand_Category']].head())
```

```python
df['Orders_Per_Queue_Position'] = df['Order_Volume'] / (df['Queue_Length'] + 1)

print("Queue pressure metric 'Orders_Per_Queue_Position' created successfully.")
print("Displaying new queue pressure metric for verification:")
display(df[['Order_Volume', 'Queue_Length', 'Orders_Per_Queue_Position']].head())

print("\n--- Feature Engineering Complete ---")
print("Final DataFrame head with new features:")
display(df.head())
```

```python
staffing_pressure_threshold = df['Orders_Per_Employee'].quantile(0.75)

potential_pressure_periods = df[df['Orders_Per_Employee'] > staffing_pressure_threshold]

print(f"\nPeriods where 'Orders_Per_Employee' exceeds the 75th percentile ({staffing_pressure_threshold:.2f}):")
print(f"Number of such periods: {potential_pressure_periods.shape[0]} out of {df.shape[0]} total periods.")

print("Sample of periods showing potential staffing pressure:")
display(potential_pressure_periods[['Date', 'Outlet_ID', 'Hour', 'Order_Volume', 'Staff_Count', 'Orders_Per_Employee', 'Queue_Length', 'Average_Preparation_Time']].head())

print("\nThese periods show higher order volumes per staff member. While this *could* indicate understaffing,\n     it requires further analysis with queue length and preparation time to confirm actual stress on operations.\n     We will investigate this further in the overstaffing/understaffing analysis step.")
print("\n--- Orders Per Employee Analysis Complete ---")
```

```python
ov_75 = df['Order_Volume'].quantile(0.75)
ql_75 = df['Queue_Length'].quantile(0.75)
apt_75 = df['Average_Preparation_Time'].quantile(0.75)
ope_75 = df['Orders_Per_Employee'].quantile(0.75)

ov_25 = df['Order_Volume'].quantile(0.25)
ql_25 = df['Queue_Length'].quantile(0.25)
ope_25 = df['Orders_Per_Employee'].quantile(0.25)

staff_median = df['Staff_Count'].median()

print("Calculated thresholds for staffing status:")
print(f"  Order_Volume (High): > {ov_75:.2f}")
print(f"  Queue_Length (High): > {ql_75:.2f}")
print(f"  Average_Preparation_Time (High): > {apt_75:.2f}")
print(f"  Orders_Per_Employee (High): > {ope_75:.2f}")
print(f"  Order_Volume (Low): < {ov_25:.2f}")
print(f"  Queue_Length (Low): < {ql_25:.2f}")
print(f"  Orders_Per_Employee (Low): < {ope_25:.2f}")
print(f"  Median Staff_Count: {staff_median:.2f}")
```

```python
def classify_staffing_status(row):
    is_high_demand = row['Order_Volume'] > ov_75
    is_high_queue = row['Queue_Length'] > ql_75
    is_high_prep_time = row['Average_Preparation_Time'] > apt_75
    is_high_ope = row['Orders_Per_Employee'] > ope_75

    if (is_high_demand and is_high_queue and is_high_prep_time) or \
       (is_high_demand and is_high_ope and is_high_prep_time) or \
       (is_high_queue and is_high_ope and is_high_prep_time):
        return 'Understaffed'

    is_low_demand = row['Order_Volume'] < ov_25
    is_low_queue = row['Queue_Length'] < ql_25
    is_low_ope = row['Orders_Per_Employee'] < ope_25
    is_high_staff_for_low_demand = row['Staff_Count'] >= staff_median

    if (is_low_demand and is_low_queue and is_low_ope) or \
       (is_low_demand and is_high_staff_for_low_demand and is_low_ope):
        return 'Potentially Overstaffed'

    return 'Balanced'

df['Staffing_Status'] = df.apply(classify_staffing_status, axis=1)

print("Staffing_Status column created successfully.")
print("Distribution of Staffing_Status:")
display(df['Staffing_Status'].value_counts())

print("\nSample of data with Staffing_Status:")
display(df[['Order_Volume', 'Queue_Length', 'Average_Preparation_Time', 'Staff_Count', 'Orders_Per_Employee', 'Staffing_Status']].sample(5))

print("\n--- Overstaffing and Understaffing Analysis Complete ---")
```

```python
from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt

def plot_learning_curve(estimator, title, X, y, axes=None, ylim=None, cv=None,
                        n_jobs=None, train_sizes=np.linspace(.1, 1.0, 5)):

    if axes is None:
        _, axes = plt.subplots(1, 3, figsize=(20, 5))

    axes[0].set_title(title)
    if ylim is not None:
        axes[0].set_ylim(*ylim)
    axes[0].set_xlabel("Training examples")
    axes[0].set_ylabel("Score")

    train_sizes, train_scores, test_scores, fit_times, _ = \
        learning_curve(estimator, X, y, cv=cv, n_jobs=n_jobs,
                       train_sizes=train_sizes, return_times=True)
    train_scores_mean = np.mean(train_scores, axis=1)
    train_scores_std = np.std(train_scores, axis=1)
    test_scores_mean = np.mean(test_scores, axis=1)
    test_scores_std = np.std(test_scores, axis=1)
    fit_times_mean = np.mean(fit_times, axis=1)
    fit_times_std = np.std(fit_times, axis=1)

    axes[0].grid()
    axes[0].fill_between(train_sizes, train_scores_mean - train_scores_std,
                         train_scores_mean + train_scores_std, alpha=0.1,
                         color="r")
    axes[0].fill_between(train_sizes, test_scores_mean - test_scores_std,
                         test_scores_mean + test_scores_std, alpha=0.1,
                         color="g")
    axes[0].plot(train_sizes, train_scores_mean, 'o-', color="r",
                 label="Training score")
    axes[0].plot(train_sizes, test_scores_mean, 'o-', color="g",
                 label="Cross-validation score")
    axes[0].legend(loc="best")

    axes[1].grid()
    axes[1].plot(train_sizes, fit_times_mean, 'o-')
    axes[1].fill_between(train_sizes, fit_times_mean - fit_times_std,
                         fit_times_mean + fit_times_std, alpha=0.1)
    axes[1].set_xlabel("Training examples")
    axes[1].set_ylabel("fit_times")
    axes[1].set_title("Scalability of the model")

    axes[2].grid()
    axes[2].plot(fit_times_mean, test_scores_mean, 'o-')
    axes[2].fill_between(fit_times_mean, test_scores_mean - test_scores_std,
                         test_scores_mean + test_scores_std, alpha=0.1)
    axes[2].set_xlabel("fit_times")
    axes[2].set_ylabel("Score")
    axes[2].set_title("Performance of the model")

    return plt

print("Generating learning curves...")

fig, axes = plt.subplots(1, 3, figsize=(20, 5))

from sklearn.model_selection import StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

plot_learning_curve(rf_classifier, "Learning Curves (Random Forest)", X, y,
                    axes=axes, ylim=(0.5, 1.01), cv=skf, n_jobs=-1)

plt.show()

print("\n--- Learning Curves Generated ---")
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import LabelBinarizer
import matplotlib.pyplot as plt
import numpy as np

print("Generating Multi-Class ROC AUC Curves...")

label_binarizer = LabelBinarizer()
y_test_binarized = label_binarizer.fit_transform(y_test)
n_classes = y_test_binarized.shape[1]

y_score = rf_classifier.predict_proba(X_test)

fpr = dict()
tpr = dict()
roc_auc = dict()
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_binarized[:, i], y_score[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

plt.figure(figsize=(10, 8))
colors = plt.cm.get_cmap('viridis', n_classes)
for i, class_name in enumerate(label_binarizer.classes_):
    plt.plot(fpr[i], tpr[i], color=colors(i), lw=2,
             label=f'ROC curve of class {class_name} (area = {roc_auc[i]:0.2f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Multi-Class ROC Curves for Random Forest Classifier (One-vs-Rest)')
plt.legend(loc="lower right")
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print("\n--- Multi-Class ROC AUC Curves Generated ---")
```

```python
print("\n--- Outlet Comparison Summary ---")

metrics_for_comparison = [
    'Order_Volume',
    'Queue_Length',
    'Average_Preparation_Time',
    'Staff_Count',
    'Orders_Per_Employee'
]

outlet_summary = df.groupby('Outlet_ID')[metrics_for_comparison].mean()

print("Average metrics by Outlet_ID:")
display(outlet_summary)

print("\nBased on these metrics, we can infer which outlets experience greater demand pressure.")
print("For example, an outlet with higher average Order_Volume, Queue_Length, and Orders_Per_Employee")
print("likely faces more demand pressure. Conversely, lower values might indicate less pressure or different operational patterns.")

ranks_order_volume = outlet_summary['Order_Volume'].rank(ascending=False)
ranks_queue_length = outlet_summary['Queue_Length'].rank(ascending=False)
ranks_prep_time = outlet_summary['Average_Preparation_Time'].rank(ascending=False)
ranks_ope = outlet_summary['Orders_Per_Employee'].rank(ascending=False)

outlet_pressure_rank = pd.DataFrame({
    'Order_Volume_Rank': ranks_order_volume,
    'Queue_Length_Rank': ranks_queue_length,
    'Average_Preparation_Time_Rank': ranks_prep_time,
    'Orders_Per_Employee_Rank': ranks_ope
})

outlet_pressure_rank['Overall_Pressure_Score'] = outlet_pressure_rank.sum(axis=1)
outlet_pressure_rank = outlet_pressure_rank.sort_values(by='Overall_Pressure_Score')

print("\nOutlet Demand Pressure Ranking (lower score = higher pressure):")
display(outlet_pressure_rank)

print("\nFrom this ranking, outlets with lower 'Overall_Pressure_Score' are experiencing higher demand pressure based on the combined average metrics.")

print("\n--- Outlet Comparison Complete ---")
```
